## Arrays

Numpy arrays are similar to python lists, with key differences:
- NumPy arrays store elements of same(Homogeneous) datatype, list can store multiple data types

The NumPy array has benifts like:
- Numpy Arrays are more memory effiecient.
- They are much faster.
- Many buit-in funcitons. 
- They are desined to perform numerical operations on largedatasets.

```python
arr = np.array([1, 2, 3])
arr
```

## Basic NumPy Methods/APIs on Arrays

- Creating a 2d array

```python
TwoD_array = np.array([[1, 2, 3], [4, 5, 6]])
TwoD_array  
```

```python
# Access any element in the array using its index.
TwoD_array[1][2]  # Accessing the third element of the second row
```

- Check data types

```python
TwoD_array.dtype
```

- explicitly define the data type of array

```python
newArr = np.array([1, 2, 3], dtype=np.float64)
```

- Find length/size of array

```python
newArr.size
```

- Get rows and columns i.e shape


```python
TwoD_array.shape
```

- Sort an array 

```python
TwoD_array.sort() # Axis is by default 1, so it sorts the array row wise
TwoD_array.sort(axis=0) # sorting the array column wise

# The above is used for in-place sorting.

# We can use np.sort() to return sorted array without modifying the original array.
np.sort(TwoD_array, axis=1) # sorting row wise
np.sort(TwoD_array, axis=0) # sorting column wise

np.sort(TwoD_array, axis=None) # sorting the entire array as if it were a 1D array
```

- Initialize a new array with zeros

```python
np.zeros(5) # Creates a 1D array of zeros with 5 elements
np.zeros((2, 3)) # Creates a 2D array of zeros with 2 rows and 3 columns
```

- Intialize new array with ones 

```python
np.ones(4) # Creates a 1D array of ones with 4 elements
np.ones((3, 2)) # Creates a 2D array of ones with 3 rows and 2 columns
```

- Initialize a new array with any value

```python
np.full((2, 3), 7) # Creates a 2D array of shape (2, 3) filled with the value 7
```

- Some other methods to initialize

```python
np.arange(0, 10) # Creates a 1D array with values from 0 to 9

np.arange(0, 10, 2) # Creates a 1D array with values from 0 to 9 with a step of 2

np.linspace(1, 5, 5) # Creates a 1D array with 5 evenly spaced values between 1 and 5
```

- Change the shape of an array

```python
TwoD_array.reshape(3, 2) # Reshaping the array to have 3 rows and 2 columns
```

- Find min or max

```python
TwoD_array.min() # Returns the minimum value in the array
TwoD_array.max() # Returns the maximum value in the array
```

- Calculate Sum

```python
TwoD_array.sum() # Returns the sum of all elements in the array
```

```python
TwoD_array.sum(axis=0) # Returns the sum of each column
```

```python
TwoD_array.sum(axis=1) # Returns the sum of each row
```

- Iterate through an array

```python
for row in TwoD_array:
    print(row)
```

- Square of all elements on array

```python
np.square(TwoD_array) # Returns the element-wise square of the array
```

- Square root of all element

```python
np.sqrt(TwoD_array) # Returns the element-wise square root of the array
```

There are many more methods/api from numPy, look up docs for more

## Matrix Manipulation

A large amount of data we work with is in the form of tables i. e rows and cols, this is where matrix manipulation comes handy

```python
# Lets conside 2 2d arrays represention q1 and q1 sales of a company.

q1_sales = np.array([[100, 200, 300], [400, 500, 600]])

q2_sales = np.array([[150, 250, 350], [450, 550, 650]])
```

- Matrix addition

```python
# We can perform element-wise addition to get the total sales for each quarter.
total_sales = q1_sales + q2_sales
total_sales
```
