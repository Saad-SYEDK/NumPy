## Arrays

Numpy arrays are similar to python lists, with key differences:
- NumPy arrays store elements of same(Homogeneous) datatype, list can store multiple data types

The NumPy array has benifts like:
- Numpy Arrays are more memory effiecient.
- They are much faster.
- Many buit-in funcitons. 
- They are desined to perform numerical operations on largedatasets.

```python
arr = np.array([1, 2, 3])
arr
```

## Basic Operations on Arrays

- Creating a 2d array

```python
TwoD_array = np.array([[1, 2, 3], [4, 5, 6]])
TwoD_array  
```

```python
# Access any element in the array using its index.
TwoD_array[1][2]  # Accessing the third element of the second row
```

- Check data types

```python
TwoD_array.dtype
```

- explicitly define the data type of array

```python
newArr = np.array([1, 2, 3], dtype=np.float64)
```

- Find length/size of array

```python
newArr.size
```

- Get rows and columns i.e shape


```python
TwoD_array.shape
```

- Sort an array 

```python
TwoD_array.sort() # Axis is by default 1, so it sorts the array row wise
TwoD_array.sort(axis=0) # sorting the array column wise

# The above is used for in-place sorting.

# We can use np.sort() to return sorted array without modifying the original array.
np.sort(TwoD_array, axis=1) # sorting row wise
np.sort(TwoD_array, axis=0) # sorting column wise

np.sort(TwoD_array, axis=None) # sorting the entire array as if it were a 1D array
```
